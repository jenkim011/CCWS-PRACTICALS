```python
y=df['Diagnosis']
X=df.drop(['Diagnosis'], axis=1)
```

```python
classifier=CategoricalNB()
classifier.fit(X,y)
```

```python
classifier=MultinomialNB()
classifier.fit(X,y)
```

```python
classifier=GaussianNB()
classifier.fit(X,y)
```

```python
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.6, random_state=42)
```

```python
classifier=MultinomialNB()
classifier.fit(X_train,y_train)
```

```python
prediction=classifier.predict(X_test)
```

```python
print("accuracy_score",accuracy_score(y_test,prediction))
print("precision : ", precision_score(y_test,prediction,average='weighted'))    
print("recall " , recall_score(y_test,prediction,average='weighted'))
print("f1-score ",f1_score (y_test,prediction,average='weighted'))
print("confusion_matrix: \n",confusion_matrix(y_test,prediction))
print("classification_report \n",classification_report(y_test,prediction))
```
